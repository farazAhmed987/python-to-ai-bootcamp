## loading dataset

## Data preprocessing

Create a cleaned copy of the original dataset. This keeps the original `df` unchanged so the earlier missing-value checks can still show the raw data.

```python
# Make a copy so the original dataset remains unchanged
df_clean = df.copy()

# Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())

# Remove duplicate rows, if any
df_clean = df_clean.drop_duplicates()

# Check missing values before cleaning
print("\nMissing values before preprocessing:")
display(df_clean.isnull().sum())
```

```python
# Fill missing ages with the median age
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())

# Fill missing embarkation ports with the most common value
df_clean["Embarked"] = df_clean["Embarked"].fillna(df_clean["Embarked"].mode()[0])

# Cabin has many missing values, so remove this column for this simple analysis
df_clean = df_clean.drop(columns=["Cabin"])

# Standardize text values by removing extra spaces and using consistent case
df_clean["Sex"] = df_clean["Sex"].str.strip().str.lower()
df_clean["Embarked"] = df_clean["Embarked"].str.strip().str.upper()

# Create useful family-related features
df_clean["FamilySize"] = df_clean["SibSp"] + df_clean["Parch"] + 1
df_clean["IsAlone"] = (df_clean["FamilySize"] == 1).astype(int)

print("Preprocessing completed successfully!")
```

```python
# Review the cleaned dataset
print("Cleaned dataset shape:", df_clean.shape)
print("\nMissing values after preprocessing:")
display(df_clean.isnull().sum())

print("\nPreview of cleaned data:")
display(df_clean.head())

# Save the cleaned dataset as a separate CSV file
df_clean.to_csv("titanic_cleaned.csv", index=False)
print("\nCleaned dataset saved as titanic_cleaned.csv")
```
